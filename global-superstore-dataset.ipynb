```python
numeric_cols = df.select_dtypes(include="number").columns

fig, axes = plt.subplots(
    nrows=1, ncols=len(numeric_cols),
    figsize=(4 * len(numeric_cols), 5)
)

for ax, col in zip(axes, numeric_cols):
    sns.boxplot(y=df[col], ax=ax, color="steelblue",
                flierprops=dict(marker="o", color="red",
                                markersize=5, linestyle="none"))
    ax.set_title(col)

fig.suptitle("Outlier Check – All Numeric Columns", fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
```

```python
cat_cols = df.select_dtypes(include=["object", "category"]).columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

## Handle outliers ##

```python
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)
```

```python
numeric_cols = df.select_dtypes(include="number").columns

fig, axes = plt.subplots(
    nrows=1, ncols=len(numeric_cols),
    figsize=(4 * len(numeric_cols), 5)
)

for ax, col in zip(axes, numeric_cols):
    sns.boxplot(y=df[col], ax=ax, color="steelblue",
                flierprops=dict(marker="o", color="red",
                                markersize=5, linestyle="none"))
    ax.set_title(col)

fig.suptitle("Outlier Check – All Numeric Columns", fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

# Top Selling Categories

```python
top_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
top_category
```

```python
import plotly.express as px

px.bar(top_category.reset_index(),
       x="Category",
       y="Sales",
       title="Top Selling Categories")
```

# Top of the Dolouk

```python
top_country = (
    df.groupby("Country")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

top_country
```

```python
fig = px.bar(
    top_country,
    x="Sales",
    y="Country",
    orientation="h",
    title="Top 10 Countries by Sales",
    text="Sales"
)

fig.update_layout(
    yaxis={'categoryorder':'total ascending'},
    title_x=0.5
)

fig.show()
```

# The effect of discounts on sales

```python
df.groupby("Discount")["Sales"].mean()
```

```python
avg_sales_discount = (
    df.groupby("Discount")["Sales"]
      .mean()
      .reset_index()
)

fig = px.line(
    avg_sales_discount,
    x="Discount",
    y="Sales",
    markers=True,
    title="Average Sales by Discount Level"
)

fig.update_layout(title_x=0.5)
fig.show()
```

# Sales over the years

```python
sales_by_year = df.groupby("Year")["Sales"].sum().reset_index()

px.line(sales_by_year,
        x="Year",
        y="Sales",
        title="Sales Trend Over Years")
```

# Weekly analysis

```python
weekly_sales = df.groupby("weeknum")["Sales"].sum().reset_index()

px.line(weekly_sales,
        x="weeknum",
        y="Sales",
        title="Weekly Sales Trend")
```

# Best Customers

```python
top_customers = df.groupby("Customer.Name")["Sales"].sum().sort_values(ascending=False).head(5)
top_customers
```

```python
top_customers = (
    df.groupby("Customer.Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
      .reset_index()
)


fig = px.bar(
    top_customers,
    x="Sales",
    y="Customer.Name",
    orientation="h",
    text="Sales",
    title="Top 5 Customers by Sales"
)


fig.update_layout(
    yaxis={'categoryorder':'total ascending'},
    title_x=0.5
)

fig.show()
```

# Average Shipping Days by Ship Mode

```python
df["Order.Date"] = pd.to_datetime(df["Order.Date"])
df["Ship.Date"] = pd.to_datetime(df["Ship.Date"])

df["Shipping_Days"] = (df["Ship.Date"] - df["Order.Date"]).dt.days

shipping_avg = df.groupby("Ship.Mode")["Shipping_Days"].mean().reset_index()

fig6 = px.bar(
    shipping_avg,
    x="Ship.Mode",
    y="Shipping_Days",
    title="Average Shipping Days by Ship Mode"
)

fig6.update_layout(title_x=0.5)
fig6.show()
```

# Sales by Segment

```python
segment_sales = df.groupby("Segment")["Sales"].sum().reset_index()

fig7 = px.pie(
    segment_sales,
    names="Segment",
    values="Sales",
    title="Sales by Customer Segment"
)

fig7.update_layout(title_x=0.5)
fig7.show()
```
